```python
import gymnasium as gym
from gymnasium.envs.registration import register
import pulsrEnv   # must import the file that defines PulsrEnv

# Clear old registrations (useful in notebooks)
if "pulsr-virtualenv-v0" in gym.envs.registration.registry:
    del gym.envs.registration.registry["pulsr-virtualenv-v0"]

register(
    id="pulsr-virtualenv-v0",
    entry_point="pulsrEnv:PulsrEnv",  # module_name:class_name
)

env = gym.make("pulsr-virtualenv-v0")
print("Env created:", env)
```

```python
from gymnasium.utils.env_checker import check_env

# Create an instance of the registered environment
env = gym.make("pulsr-virtualenv-v0")

# This will catch many common issues
try:
    check_env(env.unwrapped)
    # check_env(env)
    print("Environment passes all checks!")
except Exception as e:
    print(f"Environment has issues: {e}")
    # print(e)
```

### The Virtual Puslr Environment is confirmed to conform with the general guidelines of gymnasium environments

```python
# Test with Known Actions without input trajectories

# Test specific action sequences to verify behavior
env = gym.make("pulsr-virtualenv-v0")
obs, info = env.reset(seed=42)  # Use seed for reproducible testing

obs_agent = obs[0], obs[1]
obs_target = obs[2], obs[3]

print(f"Starting position - Agent: {obs_agent}, Target: {obs_target}")

# Test each action type
actions = [0, 1, 2, 3]  # right, up, left, down
for action in actions:
    old_pos = list(obs_agent).copy()
    obs, reward, terminated, truncated, info = env.step(action)
    new_pos = obs[0], obs[1]
    print(f"Action {action}: {old_pos} -> {new_pos}, reward={reward}")
```
