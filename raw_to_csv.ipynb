```python
import csv

# Input and output file paths
input_file = "software/Debug/sessions_files/raw_origin_from_br_corner.txt"
output_file = "software/Debug/sessions_files/raw_origin_from_br_corner.csv"

```

```python
# Open the raw text file
with open(input_file, "r") as infile, open(output_file, "w", newline="") as outfile:
    writer = csv.writer(outfile)
    writer.writerow(["angle1", "angle2", "force1", "force2"])  # CSV header
    
    lines = infile.readlines()
    i = 0
    while i < len(lines):
        line = lines[i].strip()
        
        if line.startswith("angles:"):
            # Extract angles (take first occurrence, skip duplicate)
            angles = line.replace("angles:", "").strip()
            angle1, angle2 = [int(x.strip()) for x in angles.split(",")]
            
            # Skip the next duplicate angles line
            i += 2  
            
            # Get force line
            force_line = lines[i].strip()
            if force_line.startswith("force:"):
                force = force_line.replace("force:", "").strip()
                force1, force2 = [int(x.strip()) for x in force.split(",")]
                
                # Write one clean row to CSV
                writer.writerow([angle1, angle2, force1, force2])
        
        i += 1

print(f"✅ Data has been successfully converted to {output_file}")
```

```python
import csv
import glob
import os

# Folder containing your text files
input_folder = "software/Debug/sessions_files/"   # change this to your folder path, e.g., "data/"

# Get all .txt files in the folder
text_files = glob.glob(os.path.join(input_folder, "*.txt"))

for input_file in text_files:
    # Create output file with same base name but .csv extension
    base_name = os.path.splitext(os.path.basename(input_file))[0]
    output_file = os.path.join(input_folder, base_name + ".csv")
    
    with open(input_file, "r") as infile, open(output_file, "w", newline="") as outfile:
        writer = csv.writer(outfile)
        writer.writerow(["angle1", "angle2", "force1", "force2"])  # CSV header

        lines = infile.readlines()
        i = 0
        while i < len(lines):
            line = lines[i].strip()

            if line.startswith("angles:"):
                # Extract angles (first line only, skip duplicate)
                angles = line.replace("angles:", "").strip()
                angle1, angle2 = [int(x.strip()) for x in angles.split(",")]

                # Make sure we don’t go out of range
                if i + 2 < len(lines):
                    i += 2
                    force_line = lines[i].strip()

                    if force_line.startswith("force:"):
                        force = force_line.replace("force:", "").strip()
                        force1, force2 = [int(x.strip()) for x in force.split(",")]

                        # Write row to CSV
                        writer.writerow([angle1, angle2, force1, force2])
                else:
                    break  # no more lines to read safely

            i += 1
    
    print(f"✅ Converted {input_file} → {output_file}")
```
