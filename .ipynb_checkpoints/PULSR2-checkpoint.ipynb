```python
# Scan all available serial ports on the computer.
# Filter the list to include only ports whose description contains "Arduino".
# Store the device name (e.g., "COM3") in the comPort list.

# Read comports from usb_ser.txt into a list
with open("usb_ser.txt", "r") as f:
    usb_ser_ports = [line.strip() for line in f if line.strip()]

comPorts = []

for p in serial.tools.list_ports.comports():
    # check if the device is listed in usb_ser.txt
    if p.device in usb_ser_ports:
        comPorts.append(p.device)

print("Final comPorts:", comPorts)


# Opens a serial connection to the first detected Arduino port at a baud rate of 2,000,000.
# Prepares the system for sending commands and receiving data from the Arduino.
# pulsr2.initialize_communication(2000000,comPort[0])	
```

```python
import serial
from pulsr2API import pulsr
import time

# Initialize pulsr object for sending control signals on COM14
pulsr_control = pulsr()
pulsr_control.initialize_communication(2000000, 'COM14')
print("Initialized control communication on COM14.")

# Initialize serial connections for receiving data on COM12 and COM13
arduino_rx1 = serial.Serial(port='COM12', baudrate=2000000, timeout=1)
arduino_rx2 = serial.Serial(port='COM13', baudrate=2000000, timeout=1)
print("Initialized data receiving on COM12 and COM13.")

# Example: Check communication status
if pulsr_control.check_communication():
    print("COM14 (control) is open.")
else:
    print("COM14 (control) failed to open.")

if arduino_rx1.isOpen():
    print("COM12 (receive) is open.")
else:
    print("COM12 (receive) failed to open.")

if arduino_rx2.isOpen():
    print("COM13 (receive) is open.")
else:
    print("COM13 (receive) failed to open.")

# Example: Read data from COM12 and COM13
try:
    while True:
        if arduino_rx1.in_waiting:
            data1 = arduino_rx1.readline().decode('utf-8', errors='ignore').strip()
            print(f"COM12 received: {data1}")
        if arduino_rx2.in_waiting:
            data2 = arduino_rx2.readline().decode('utf-8', errors='ignore').strip()
            print(f"COM13 received: {data2}")
        time.sleep(0.1)
except KeyboardInterrupt:
    print("Exiting...")

# Close all ports on exit
arduino_rx1.close()
arduino_rx2.close()
pulsr_control.close_communication()

print("Communications closed")
```

```python
#  creates a window of the specified size
starting_dialog_box=pygame.display.set_mode(starting_dialog_box_size)

# The code assigns this window to three different variables: starting_dialog_box, background, and screen.
# Each variable is intended to represent a different layer or purpose for drawing (dialog box, background, and main screen).
# However, in Pygame, calling set_mode multiple times will actually recreate the window each time,
# so only the last assignment (screen) will be the active display window.
background=pygame.display.set_mode(starting_dialog_box_size)	#foundation surface definition
screen=pygame.display.set_mode(starting_dialog_box_size)	#starting dialog box surface definition
```

```python
#caption and text defineition for starting dialog box

# Set the title of the Pygame window to "PULSR2.0"
#make caption of current display which is screen-> starting_dialog_box_size
pygame.display.set_caption('PULSR2.0')	 

# Create three font objects with different sizes for rendering text.
oaufont = pygame.font.SysFont('Arial', 25)
oaufont2 = pygame.font.SysFont('Arial', 15)
oaufont3 = pygame.font.SysFont('Arial', 10)

# Prepare text surfaces with specific messages and colors (black)
meaningtext = oaufont.render('Platform for Upper Limb Stroke Rehabilitation', False, (0, 0, 0))
oautext = oaufont2.render('Obafemi Awolowo University, Ile-Ife', False, (0, 0, 0))
dialog_message = oaufont3.render('PULSR2 starting up.........', False, (0, 0, 0))

# Creates a clock object to control the frame rate of the Pygame window
clock=pygame.time.Clock()
```

```python
# Set the entire dialog box background to white
background.fill((255,255,255))

# Place three text surfaces (meaningtext, oautext, dialog_message) at specified positions
# on the dialog box
screen.blit(meaningtext,(10,60))
screen.blit(oautext,(135,90))
screen.blit(dialog_message,(7,170))

# Refreshes the window to show the new background and text
pygame.display.update()

# Limits the frame rate to 20 frames per second and pauses for 1 second.
clock.tick(20)
time.sleep(1)
```

```python
'''
	code to display pulsr2 starting small dialog box
		i.check communication status and report on dialog box
'''
#i.check pulsr2 communication status and report on dialog box
if pulsr2.check_communication()==True:
	dialog_message = oaufont3.render('communication port active', False, (0, 0, 0))
	background.fill((255,255,255))	#fill starting dialog box background with white
	screen.blit(meaningtext,(10,60))
	screen.blit(oautext,(135,90))
	screen.blit(dialog_message,(7,170))
	pygame.display.update()
	clock.tick(20)
	time.sleep(1)
else:
	dialog_message = oaufont3.render('communication port not active', False, (0, 0, 0))
	background.fill((255,255,255))	#fill dialog box background with white
	# background.blit(image,(0,0))
	screen.blit(meaningtext,(10,60))
	screen.blit(oautext,(135,90))
	screen.blit(dialog_message,(7,170))
	pygame.display.update()
	clock.tick(20)
	time.sleep(1)
	pygame.quit()
	quit()
```
