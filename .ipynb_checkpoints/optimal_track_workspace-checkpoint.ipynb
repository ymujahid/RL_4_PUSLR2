```python
# firstly concatenate all x and all y 

import pandas as pd
import glob
import os

# Path to the parent folder containing subfolders of csv files
bases = ["track_trajectory_coordinates", "workspace_coordinates"]

# Get all csv files from subfolders

workspace_dict = {}
track_dict = {}
for base in bases:
    csv_files = glob.glob(os.path.join(base, "**", "*.csv"), recursive=True)
    for file in csv_files:
        parent = os.path.dirname(file)
        filename = os.path.splitext(os.path.basename(file))[0]  # csv name without extension
        if base == bases[0]:
            track_dict[filename] = pd.read_csv(file)
        elif base == bases[1]:
            workspace_dict[filename] = pd.read_csv(file)
        else:
            print("error")




# Example: access one DataFrame
print(workspace_dict[list(workspace_dict.keys())[0]].head())
print(list(track_dict.keys()))
print(list(workspace_dict.keys()))

```

```python
# Ensure they all have same number of rows
n_instances = len(workspace_dict[list(workspace_dict.keys())[0]])

n_instances
```

```python
# Concatenate x and y values instance-wise for Track Trajectory Coordinates
track_concat = {
    "instance": list(range(n_instances)),
    "x_values": [],
    "y_values": []
}

for i in range(n_instances):
    xs = [df.iloc[i]['x'] for name,df in track_dict.items()]  # gather x at row i
    ys = [df.iloc[i]['y'] for name,df in track_dict.items()]  # gather y at row i
    track_concat["x_values"].append(xs)
    track_concat["y_values"].append(ys)

# Final DataFrame
track_concat_df = pd.DataFrame(track_concat)
track_concat_df.head()
```

```python
# fish out the df with abnormal length
for name,df in workspace_dict.items():
    print(len(df))
```

```python
# Concatenate x and y values instance-wise for Workspace Trajectory Coordinates
workspace_concat = {
    "instance": list(range(n_instances)),
    "x_values": [],
    "y_values": []
}

for i in range(n_instances):
    xs = [df.iloc[i]['x'] for name,df in workspace_dict.items() if len(df) == n_instances]  # gather x at row i
    ys = [df.iloc[i]['y'] for name,df in workspace_dict.items() if len(df) == n_instances]  # gather y at row i
    # print(len(xs), len(ys)) # uncomment for debugging
    workspace_concat["x_values"].append(xs)
    workspace_concat["y_values"].append(ys)

# Final DataFrame

workspace_concat_df = pd.DataFrame(workspace_concat)
workspace_concat_df.head()
```

```python
# take mean as the optimal

# Compute means for workspace
mean_x = workspace_concat_df["x_values"].apply(lambda xs: sum(xs)/len(xs))
mean_y = workspace_concat_df["y_values"].apply(lambda ys: sum(ys)/len(ys))

# Store in a new DataFrame
workspace_mean_df = pd.DataFrame({
    "instance": workspace_concat_df["instance"],
    "mean_x": mean_x,
    "mean_y": mean_y
})

print(workspace_mean_df.head())
print(workspace_mean_df.shape)
```

```python
# Compute means for track
mean_x = track_concat_df["x_values"].apply(lambda xs: sum(xs)/len(xs))
mean_y = track_concat_df["y_values"].apply(lambda ys: sum(ys)/len(ys))

# Store in a new DataFrame
track_mean_df = pd.DataFrame({
    "instance": track_concat_df["instance"],
    "mean_x": mean_x,
    "mean_y": mean_y
})

print(track_mean_df.head())
print(track_mean_df.shape)
```

```python
# visualize coordinates

import matplotlib.pyplot as plt

def plot_trajectory(df, x_col='x', y_col='y', title="the Dataframe"):
    """
    Plots the trajectory of the end effector from its (x, y) coordinates in df.
    
    Parameters:
    - df: pandas DataFrame containing x and y columns
    - x_col: name of the x-coordinate column
    - y_col: name of the y-coordinate column
    """
    x = df[x_col]
    y = df[y_col]
    
    plt.figure(figsize=(8, 6))
    plt.plot(x, y, marker='o', markersize=2, linewidth=1, alpha=0.7, label="Trajectory")
    
    # Highlight start and end points
    plt.scatter(x.iloc[0], y.iloc[0], color="green", s=100, label="Start")
    plt.scatter(x.iloc[-1], y.iloc[-1], color="red", s=100, label="End")
    
    plt.xlabel("X Position")
    plt.ylabel("Y Position")
    plt.title(f"End Effector Trajectory for {title}")
    plt.legend()
    plt.axis("equal")  # keeps aspect ratio consistent
    plt.grid(True)
    plt.show()
```

```python
plot_trajectory(track_mean_df, 'mean_x', 'mean_y', "The Mean Circular Track")
```

```python
plot_trajectory(workspace_mean_df, 'mean_x', 'mean_y', "The Mean Workspace")
```

```python
# the Median seems to be a better representation of the workspace trajectory but no conclusion for the track trajectory

# fit a circle into the plot
import numpy as np
from scipy.optimize import least_squares

def fit_circle(x, y):
    """Fit a circle to (x, y) points and return center, radius, residuals"""
    def calc_R(xc, yc):
        return np.sqrt((x - xc)**2 + (y - yc)**2)

    def f(c):
        Ri = calc_R(*c)
        return Ri - Ri.mean()

    # Initial guess = centroid of points
    x_m, y_m = np.mean(x), np.mean(y)
    result = least_squares(f, (x_m, y_m))

    xc, yc = result.x
    R = np.mean(calc_R(xc, yc))
    residuals = np.std(calc_R(xc, yc) - R)

    return (xc, yc), R, residuals

# Example usage:
(xc, yc), R, err = fit_circle(track_mean_df["mean_x"].values, track_mean_df["mean_y"].values)
print(f"Center: ({xc:.2f}, {yc:.2f}), Radius: {R:.2f}, Residual error: {err:.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Circle parameters
xc, yc = 4.42, 54.52
R = 9.48
residual_error = 0.756

# Create circle points
theta = np.linspace(0, 2*np.pi, 500)
x_circle = xc + R * np.cos(theta)
y_circle = yc + R * np.sin(theta)

# Plot
plt.figure(figsize=(6,6))
plt.plot(x_circle, y_circle, 'b-', label=f'Fitted Circle (R={R:.2f})')
plt.scatter(xc, yc, c='red', marker='x', s=100, label='Center')
plt.gca().set_aspect('equal', adjustable='box')

plt.title("Fitted Circle")
plt.xlabel("X Position")
plt.ylabel("Y Position")
plt.legend()
plt.grid(True)
plt.show()

print(f"Center: ({xc}, {yc}), Radius: {R}, Residual Error: {residual_error}")
```

```python
import numpy as np
import pandas as pd

# Circle parameters
xc, yc = 4.42, 54.52
R = 9.48

# Desired starting point (approximate)
x_start, y_start = 9, 45

# Compute angle of starting point relative to center
theta0 = np.arctan2(y_start - yc, x_start - xc)

# Generate points
n_points = 239
theta = np.linspace(theta0, theta0 + 2*np.pi, n_points)

x_circle = xc + R * np.cos(theta)
y_circle = yc + R * np.sin(theta)

# Store in DataFrame
ideal_circle_df = pd.DataFrame({
    "x": x_circle,
    "y": y_circle
})

print(ideal_circle_df.head())
print(ideal_circle_df.tail())
print(ideal_circle_df.shape)  # should be (239, 2)
```

```python
plot_trajectory(ideal_circle_df, "x", "y")
```

```python
# Save all trajectory DataFrames as CSV files
output_optimal = "Optimal_Trajectories"
os.makedirs(output_optimal, exist_ok=True)

# save ideal and optimal trajectories for circular track and workspace

names = ["ideal_circle", ]
dfs = [ideal_circle_df, ]
for i, name in enumerate(names):
    csv_path = os.path.join(output_optimal, f"{name}.csv")
    dfs[i].to_csv(csv_path, index=False)
```
