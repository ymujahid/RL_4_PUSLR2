```python
import serial.tools.list_ports

# List available ports
ports = serial.tools.list_ports.comports()
portsList = []

print("Available Ports:")
for one in ports:
    portsList.append(str(one))
    print(str(one))

print("\nPlease assign the COM ports explicitly.")

# Hardcode assignments instead of assuming order
in1_num = input("Enter COM number for Input Arduino #1: ")
in2_num = input("Enter COM number for Input Arduino #2: ")
out_num = input("Enter COM number for Output Arduino: ")

```

```python
# Build full COM strings
input_ports = ["COM" + str(in1_num), "COM" + str(in2_num)]
output_port = "COM" + str(out_num)

print("\nInput Ports:", input_ports)
print("Output Port:", output_port)

# Open serial connections
input_serials = [serial.Serial(port, 9600) for port in input_ports]
output_serial = serial.Serial(output_port, 9600)

print(input_serials, output_serial)

print("Serial connections established.")
```

```python
import time

print("Reading from input ports... Press Ctrl+C to stop.")

try:
    while True:
        for i, ser in enumerate(input_serials, start=1):
            if ser.in_waiting > 0:   # check if data is available
                line = ser.readline().decode('utf-8').strip()
                print(f"Input{i} ({ser.port}): {line}")
        
        time.sleep(0.1)  # small delay to avoid busy waiting

except KeyboardInterrupt:
    print("\nStopping...")
finally:
    # Close ports properly
    for ser in input_serials:
        ser.close()
    output_serial.close()
    print("All ports closed.")
```

```python
import serial
import time
from pulsr2API import pulsr

# --- Open sensor input ports (load cell & encoder) ---
loadcell_serial = serial.Serial('COM13', 2000000, timeout=1)
encoder_serial = serial.Serial('COM12', 2000000, timeout=1)

# --- Initialize control port via PULSR2 API ---
robot = pulsr()
robot.initialize_communication(2000000, 'COM14')
```

```python
# --- Define robot geometry and origin ---
robot.define_geometry(22, 43, 43, 65, 42)
time.sleep(2)  # let things stabilize
robot.set_origin(0, 75)

print("Setup complete. Reading sensors from COM12 & COM13, controlling motors via COM14.")

try:
    while True:
        # --- Read encoder data ---
        if encoder_serial.in_waiting > 0:
            encoder_data = encoder_serial.readline().decode('utf-8', errors='ignore').strip()
            print(f"[Encoder - COM12] {encoder_data}")

        # --- Read load cell data ---
        if loadcell_serial.in_waiting > 0:
            loadcell_data = loadcell_serial.readline().decode('utf-8', errors='ignore').strip()
            print(f"[Load Cell - COM13] {loadcell_data}")

        # --- Example control: simple demo ---
        # Move north for 1 second, then stop
        robot.motion1(1)
        robot.disable_motions()

        time.sleep(0.1)

except KeyboardInterrupt:
    print("\nStopping...")

finally:
    # --- Close all ports ---
    loadcell_serial.close()
    encoder_serial.close()
    robot.close_communication()
    print("All COM ports closed.")
```
