```python
import os
import json
from stable_baselines3 import DQN

# Path to model and artifacts folder
model_name = "dqn_pulsr_virtualenv_100000"
model_path = os.path.join("models", model_name)
artifacts_dir = "artifacts"
os.makedirs(artifacts_dir, exist_ok=True)
artifacts_path = os.path.join(artifacts_dir, f"{model_name}.json")

# Load model
model = DQN.load(model_path)

# --- Collect artifacts ---
artifacts = {}

# 1. Policy architecture
artifacts["policy_class"] = model.policy.__class__.__name__
artifacts["policy_str"] = str(model.policy)

# 2. Hyperparameters
artifacts["hyperparameters"] = {k: str(v) for k, v in model.get_parameters().items()}

# 3. Q-network weights (convert tensors to lists for JSON)
q_net_params = model.policy.q_net.state_dict()
artifacts["q_net_weights"] = {k: v.cpu().numpy().tolist() for k, v in q_net_params.items()}

# 4. Replay buffer (if exists)
try:
    model.load_replay_buffer(model_path + "_replay_buffer.pkl")
    artifacts["replay_buffer"] = "Replay buffer loaded successfully."
except Exception:
    artifacts["replay_buffer"] = "Replay buffer not found."

# --- Save to JSON ---
with open(artifacts_path, "w") as f:
    json.dump(artifacts, f, indent=4)

print(f"Artifacts saved to {artifacts_path}")
```
